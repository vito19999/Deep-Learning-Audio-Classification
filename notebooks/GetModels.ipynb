```python
import torch
import torch.nn as nn

# --- Modello 1: semplice CNN ---
class ConvNet(nn.Module):
    def __init__(self, in_channels=1, out_channels=16, kernel_size=3):
        super(ConvNet, self).__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Linear(out_channels, 10)  # esempio: 10 classi

    def forward(self, x):
        x = self.conv(x)
        x = torch.relu(x)
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x


# --- Modello 2: semplice MLP ---
class LinearNet(nn.Module):
    def __init__(self, input_size=28*28, hidden_size=128, num_classes=10):
        super(LinearNet, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        x = x.view(x.size(0), -1)  # appiattiamo l'input
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x


# --- Modello 3: modello misto conv + linear ---
class MixedNet(nn.Module):
    def __init__(self, in_channels=1, conv_out=8, fc_out=10):
        super(MixedNet, self).__init__()
        self.conv = nn.Conv2d(in_channels, conv_out, kernel_size=5)
        self.pool = nn.AdaptiveAvgPool2d(2)
        self.fc = nn.Linear(conv_out*2*2, fc_out)  # adattiamo le dimensioni dopo il pooling

    def forward(self, x):
        x = torch.relu(self.conv(x))
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x


# --- Funzione per richiamare i modelli ---
def get_model(model_name, **kwargs):
    """
    Ritorna un'istanza del modello richiesto.
    
    Args:
        model_name (str): 'conv', 'linear', 'mixed'
        **kwargs: parametri da passare al costruttore del modello
    """
    if model_name == 'conv':
        return ConvNet(**kwargs)
    elif model_name == 'linear':
        return LinearNet(**kwargs)
    elif model_name == 'mixed':
        return MixedNet(**kwargs)
    else:
        raise ValueError(f"Modello {model_name} non riconosciuto. Scegli tra 'conv', 'linear', 'mixed'.")


# --- Esempi di utilizzo ---
model1 = get_model('conv', in_channels=1, out_channels=32)
model2 = get_model('linear', input_size=784, hidden_size=256)
model3 = get_model('mixed', in_channels=3, conv_out=16, fc_out=5)

print(model1)
print(model2)
print(model3)
```
